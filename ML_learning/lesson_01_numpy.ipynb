```python
hours = [1, 2, 3, 4, 5, 6, 7]
marks = [35, 42, 50, 61, 69, 78, 88]
```

```python
import matplotlib.pyplot as plt
plt.plot(hours, marks)
plt.xlabel("marks")
plt.ylabel("score")
plt.show()
```

```python
marks = [45, 50, 52, 55, 60, 61, 65, 70, 72, 75, 80, 85, 90]

plt.hist(marks)

plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.title("Marks Distribution")

plt.show()
```

```python
plt.boxplot(marks)

plt.title("Marks Distribution")

plt.show()
```
